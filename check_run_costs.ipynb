
gpt-4o -- $2.50 / 1M input tokens    <--->      $10.00 / 1M output tokens

gpt-4o-mini -- $0.150 / 1M input tokens <--->   $0.600 / 1M output tokens


```python
split_rate = 0.15
mini_input = 0.150 
mini_output = 0.600 
max_input = 2.50 
max_output = 10.00
```

```python
with open('memory/execution_cost/token_usage.txt') as thefile:
    data = thefile.readlines()
```

```python
from ast import literal_eval
literal_eval(data[0])
```

```python
completion_tokens = 0
prompt_tokens = 0 

for line in data:
    line = literal_eval(line)
    completion_tokens += line[2]
    prompt_tokens += line[3]
```

```python
# Costs per 1M tokens for GPT-4o-mini

final_input_cost = (mini_input / 1_000_000) * prompt_tokens
final_output_cost = (mini_output / 1_000_000) * completion_tokens

print('GPT-4o-mini')
print(f"Cost per one run: Input -- {final_input_cost:.15f}, Output -- {final_output_cost:.15f}")
print(f"Number of runs to cost 1 dollar Input --  {1/final_input_cost:.2f}, Output -- {1/final_output_cost:.2f}")
```

```python
# Costs per 1M tokens for GPT-4o

final_input_cost = (max_input / 1_000_000) * prompt_tokens
final_output_cost = (max_output / 1_000_000) * completion_tokens

print('GPT-4o')
print(f"Cost per one run: Input -- {final_input_cost:.15f}, Output -- {final_output_cost:.15f}")
print(f"Number of runs to cost 1 dollar Input --  {1/final_input_cost:.2f}, Output -- {1/final_output_cost:.2f}")
```
